```python
import importlib
import os
import shutil
import site
import sys

CLEAN_PATH = "/kaggle/working/bnb_clean"

for module_name in list(sys.modules):
    if module_name == "bitsandbytes" or module_name.startswith("bitsandbytes."):
        del sys.modules[module_name]

for folder in site.getsitepackages():
    old_path = os.path.join(folder, "bitsandbytes")

    if os.path.exists(old_path):
        print("Removing conflicting package:", old_path)
        shutil.rmtree(old_path)

sys.path = [
    path for path in sys.path
    if path != CLEAN_PATH
]

sys.path.insert(0, CLEAN_PATH)
importlib.invalidate_caches()

import torch
import bitsandbytes as bnb

print("Loaded from:", bnb.__file__)
print("bitsandbytes:", bnb.__version__)
print("CUDA:", torch.version.cuda)
print("GPU available:", torch.cuda.is_available())

assert bnb.__file__.startswith(CLEAN_PATH)
assert torch.cuda.is_available()

print("IMPORT CHECK: PASS")
```
